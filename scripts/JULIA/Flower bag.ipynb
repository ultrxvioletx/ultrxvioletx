```julia
using Luxor
using Colors
```

```julia
colores = [
    RGB(0.64, 0.86, 0.75), # verde
    RGB(0.95, 0.78, 0.79), # rosa
    RGB(0.90, 0.90, 0.85), # blanco
    RGB(0.80, 0.80, 0.80), # gris claro
    RGB(0.59, 0.59, 0.59)  # gris oscuro
]

rows = 5
cols = 12

height = 80
width = 80
detuning = height/2
```

```julia
function draw_flower(x, y, center_color, petal_color)
    # pétalos
    for i in 1:6
        θ = 2π * i/6
        petal_x = x + 25 * cos(θ)
        petal_y = y + 25 * sin(θ)
        sethue(petal_color)
        circle(petal_x, petal_y, 15, :fill)
    end
    # centro
    sethue(center_color)
    circle(x, y, 10, :fill)
end
```

```julia
Drawing(cols*width + detuning, rows*height, "flores_grid.png")
background("white")
for row in 1:rows
    for col in 1:cols
        x_base = (col - 0.5) * width
        y = (row - 0.5) * height
        # desfase a las filas pares
        if row % 2 == 0
            x = x_base + detuning
        else
            x = x_base
        end
        
        center_color = rand(colores)
        petal_color = rand(colores)
        
        while petal_color == center_color
            petal_color = rand(colores)
        end
        draw_flower(x, y, center_color, petal_color)
    end
end
finish()
preview()
```
